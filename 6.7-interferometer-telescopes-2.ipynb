```python
# ============================================================================
#  CELL A3  baseline：自有信号 + baofisher 噪声（重构版）
#  - 望远镜切换：只改 main() 里的 SURVEYS 列表；全部可用配置见 TELESCOPES
#  - 命名规范：{telescope}_{mode}，无裸名；mode 后缀 = baofisher 实际使用的 mode
#  - 解析 Fisher: fisher_sum_analytic；mock Fisher: fisher_sum_mock（配 mock_file 才启用）
#  - mode 是 expt 字典的字段（与 baofisher 一致），n(x) 由 mode 自动驱动加载
#  - 输出统一口径（全部来自 bin_data / F_*）：
#      ① 每望远镜一张 7 列约束表（σ(F_ana)/σ(F_ana+F_CMB)/σ(F_ana+mock)/
#         σ(F_ana+mock+F_CMB)/σ(F_CMB)/vs_CMB%×2）+ CSV
#      ② 5 张方差图（signal/noise/analytic：realspace 方差 σ² vs z；mock 两张：k 空间 mock 方差）
#      ③ 3 张 P_noise/P_signal/SNR 图（多望远镜同图）
#      ④ 7 张信号对参数偏导图（每参数一张，多望远镜同图）
#      ⑤ 每望远镜一张 realspace voxel 方差表（var_signal/var_noise/var_total，
#         每 z-bin 一行）+ CSV/joblib + voxel_realspace_variance.png（3 子图）
#  - y 轴截距设计（科学、可复现）：
#      物理掩码：楔形/无基线区 P_noise 的巨大有限值(~10²⁰⁰)视为无效（只影响绘图曲线）
#      鲁棒窗口：所有图统一取有限值 [p2,p98] 分位 + 外扩 0.2 量级作为 ylim
#      透明化：标题标注裁剪点数；曲线全被掩码时告警
# ============================================================================

# ★ 必须在所有 matplotlib 导入之前禁用 LaTeX（系统缺少 type1cm.sty）
import matplotlib as _mpl
_mpl.rcParams["text.usetex"] = False

import warnings
from pathlib import Path
import sys
import copy
import numpy as np
import contextlib
import io
from dataclasses import dataclass, field
from typing import Tuple, List, Optional

from classy import Class

# ---- import baofisher from your repo checkout ----
BAO21CM_ROOT = "/home/liuhk/bao21cm-master/bao21cm-master"
if BAO21CM_ROOT not in sys.path:
    sys.path.insert(0, BAO21CM_ROOT)

from radiofisher import baofisher_k as bf
from radiofisher import experiments as ex

bf.KMIN_EFF = 5.0e-3    #  k 下限（单位 Mpc⁻¹）
bf.KMAX_EFF = 0.535     #  k 上限（单位 Mpc⁻¹）

C_LIGHT_KMS = 299792.458  # km/s
DELTA_C = 1.686

# baofisher 的 INF_NOISE（缺失时兜底）
BF_INF_NOISE = float(getattr(bf, "INF_NOISE", 1e200))


# ============================================================================
# 1. 望远镜分发表：按望远镜分组，每组列出模式变体
#    命名 {telescope}_{mode}；n(x) 来源标注：
#      ✅ 读文件（文件名）   配置自带 n(x) 且文件存在 → 干涉分支自动加载
#      ⚠️ 文件缺失→uniform  配置自带 n(x) 缺失 → 用 Dmin/Dmax 算均匀基线密度
#      🔢 uniform（计算）    配置无 n(x) → 干涉分支用 Dmin/Dmax 计算
#      —  不需要            单碟 / PAF / hybrid 自相关分支不读 n(x)
#    红移范围由配置的 survey_numax / survey_dnutot 换算（nu_line=1420.406 MHz）。
# ============================================================================
def normalize_mode(mode: str) -> str:
    """把 mode 字符串规范化成 baofisher.Cnoise 能正确解析的形式。

    baofisher 只在 mode[0]=='i' 时进入干涉分支；'cylinder' 不以 i 开头，
    会静默落入单碟自相关分支 → 必须翻译为 'icyl'（cylinder 干涉）。
    """
    m = str(mode).strip().lower()
    if m == "cylinder":
        return "icyl"
    allowed = {"dish", "interferom", "icyl", "iaa", "ipaf", "paf", "hybrid"}
    if m not in allowed:
        raise ValueError(f"未知 mode {mode!r}，可用: {sorted(allowed)}")
    return m


@dataclass(frozen=True)
class TelescopeSpec:
    """单条望远镜配置：experiments.py 配置对象 + 可选模式覆盖/补齐字段/mock。"""
    expt: dict                                   # experiments.py 中的配置对象（自带 mode/n(x) 等）
    extra: dict = field(default_factory=dict)    # 需补齐的字段（Dmin/Dmax、Ncyl/cyl_area）
    mock_file: Optional[str] = None              # 该条目的 mock 方差文件（配了且 z-bin 匹配才叠加）
    mode_override: Optional[str] = None          # 覆盖 mode；None=用配置自带（'cylinder' 自动→'icyl'）


TELESCOPES = {
    # ---------------- HIRAX（干涉；n(x) 文件缺失→uniform）----------------
    # 1024×6m 碟，Dmin=6/Dmax=200 为假设值；红移 z=[0.67, 2.74]
    "hirax_interferom":      TelescopeSpec(ex.HIRAX, extra={"Dmin": 6.0, "Dmax": 200.0}),
    #   ⚠️ 缺 hirax_Ndish1024_baseline1.dat → uniform（Dmin/Dmax 假设值）
    "cv_z0to3_interferom":   TelescopeSpec(ex.CVlimited_z0to3, extra={"Dmin": 6.0, "Dmax": 200.0}),
    #   ⚠️ 同 HIRAX → uniform；红移 z=[0.08, 3.25]
    "cv_z2to5_interferom":   TelescopeSpec(ex.CVlimited_z2to5, extra={"Dmin": 6.0, "Dmax": 200.0}),
    #   ⚠️ 同 HIRAX → uniform；红移 z=[1.90, 5.19]

    # ---------------- CHIME（cylinder 干涉；'cylinder'→'icyl' 自动翻译）----------------
    # 5×20m×80m 圆柱，1280 接收；红移 z=[0.67, 2.74]
    "chime_icyl":            TelescopeSpec(ex.CHIME),
    #   ✅ nx_CHIME_800.dat；自带 Ncyl=5/cyl_area=1600/Dmin/Dmax
    "chime_nocut_icyl":      TelescopeSpec(ex.CHIME_nocut),
    #   ✅ nx_CHIME_800_nocut.dat（无 FOV 截断版）
    "chime_avglow_icyl":     TelescopeSpec(ex.CHIME_avglow),
    #   ⚠️ 缺 nx_CHIME_800_avg.dat → uniform（自带 Dmin=20/Dmax=128）

    # ---------------- TIANLAI（cylinder 干涉；缺 Ncyl/cyl_area 由 extra 补）----------------
    "tianlai_icyl":          TelescopeSpec(ex.TIANLAI,      extra={"Ncyl": 8, "cyl_area": 1500.0}),
    #   ✅ nx_TIANLAI_1000.dat；红移 z=[0.39, 1.77]
    "tianlai_b2_icyl":       TelescopeSpec(ex.TIANLAIband2, extra={"Ncyl": 8, "cyl_area": 1500.0}),
    #   ✅ nx_TIANLAI_1000.dat；红移 z=[-0.10, 0.58]
    "tianlai_path_icyl":     TelescopeSpec(ex.TIANLAIpath,  extra={"Ncyl": 8, "cyl_area": 1500.0}),
    #   ✅ nx_TIANLAIpath_1000.dat；红移 z=[0.67, 1.22]

    # ---------------- SKA-MID（SKA1MID 家族；配置自带 dish，干涉需 override）----------------
    # ska_mid1: 190×15m，28K，700 MHz 带宽；红移 z=[0.25, 3.25]
    "ska_mid1_dish":         TelescopeSpec(ex.SKA1MIDbase1),                     # — 单碟自相关
    "ska_mid1_interferom":   TelescopeSpec(ex.SKA1MIDbase1, mode_override="interferom"),
    #   ✅ nx_SKAM190_dec30.dat（190 碟 baseline）
    # ska_mid2: 190×15m，20K，470 MHz 带宽；红移 z=[-0.10, 0.69]
    "ska_mid2_dish":         TelescopeSpec(ex.SKA1MIDbase2),                     # — 单碟自相关
    "ska_mid2_interferom":   TelescopeSpec(ex.SKA1MIDbase2, mode_override="interferom"),
    #   ✅ nx_SKAM190_dec30.dat
    # full 系列：254 碟（SKAM190+MeerKAT）；红移 z=[0.30, 1.64] / [-0.10, 0.69]
    "ska_midfull1_interferom": TelescopeSpec(ex.SKA1MIDfull1, mode_override="interferom"),
    "ska_midfull2_interferom": TelescopeSpec(ex.SKA1MIDfull2, mode_override="interferom"),
    #   ✅ nx_SKAMREF2_dec30.dat
    # 900/350/350upd 系列；红移 z=[-0.10,0.77] / [0.25,3.25] / [0.25,3.25]
    "ska_mid900_interferom":   TelescopeSpec(ex.SKA1MID900,   mode_override="interferom"),
    "ska_mid350_interferom":   TelescopeSpec(ex.SKA1MID350,   mode_override="interferom"),
    "ska_mid350u_interferom":  TelescopeSpec(ex.SKA1MID350upd, mode_override="interferom"),
    #   ✅ nx_SKAMREF2_dec30.dat

    # ---------------- SKA-SUR（SKA1SUR 家族；dish 与 paf 两种模式）----------------
    # SUR = 单碟 PAF（Nbeam=36，带 nu_crit）：dish 不启用 nu_crit，paf 才启用
    "ska1sur_b1_dish":       TelescopeSpec(ex.SKA1SURbase1),                     # — dish（nu_crit 不生效）
    "ska1sur_b1_paf":        TelescopeSpec(ex.SKA1SURbase1, mode_override="paf"),# — PAF（nu_crit 生效）
    #   红移 z=[0.48, 2.74]
    "ska1sur_b2_dish":       TelescopeSpec(ex.SKA1SURbase2),
    "ska1sur_b2_paf":        TelescopeSpec(ex.SKA1SURbase2, mode_override="paf"),
    #   红移 z=[0.13, 1.38]
    "ska1surfull1_dish":     TelescopeSpec(ex.SKA1SURfull1),                     # 红移 z=[0.48, 2.74]
    "ska1surfull2_dish":     TelescopeSpec(ex.SKA1SURfull2),                     # 红移 z=[0.13, 1.38]
    "ska1sur350_paf":        TelescopeSpec(ex.SKA1SUR350),                       # — 配置自带 paf；z=[0.48, 2.74]
    "ska1sur650_paf":        TelescopeSpec(ex.SKA1SUR650),                       # — 配置自带 paf；z=[0.13, 1.38]
    "ska1sur350fb_paf":      TelescopeSpec(ex.SKA1SUR350fullband),               # — 配置自带 paf；z=[0.48, 3.25]

    # ---------------- SKA-LOW（aperture-array 干涉 iaa；n(x) 缺失→uniform）----------------
    # 911/455×35m，110 MHz 临界频率；红移 z=[2.96, 4.87]（Base）/[2.96, 6.29]（Upd）/[1.74, 6.29]（Alt）
    "ska1low_iaa":           TelescopeSpec(ex.SKA1LOW, extra={"Dmin": 35.0, "Dmax": 1500.0}),
    "low_base_iaa":          TelescopeSpec(ex.LOW_Base, extra={"Dmin": 35.0, "Dmax": 1500.0}),
    "low_upd_iaa":           TelescopeSpec(ex.LOW_Upd,  extra={"Dmin": 35.0, "Dmax": 1500.0}),
    "low_alt_iaa":           TelescopeSpec(ex.LOW_Alt,  extra={"Dmin": 35.0, "Dmax": 1500.0}),
    #   ⚠️ 缺 nx_SKALOW_190*.dat → uniform（Dmin/Dmax 假设值）

    # ---------------- SKA2（iaa；自带 Dmin/Dmax）----------------
    "ska2_iaa":              TelescopeSpec(ex.SKA2),          # 🔢 uniform（Dmin=1/Dmax=1e3）；z=[0.32, 3.93]
    "ska2_mg_iaa":           TelescopeSpec(ex.SKA2_MG),       # 🔢 uniform；z=[0.32, 3.93]

    # ---------------- SKA0（早期概念）----------------
    "ska0mid_dish":          TelescopeSpec(ex.SKA0MID),       # 红移 z=[-0.10, 0.77]
    "ska0midb1_dish":        TelescopeSpec(ex.SKA0MIDB1),     # 红移 z=[0.25, 3.25]
    "ska0sur_paf":           TelescopeSpec(ex.SKA0SUR),       # 红移 z=[0.13, 1.38]

    # ---------------- MID（MG paper 系列；hybrid 自相关 或 dish）----------------
    "mid_b1_rebase_hybrid":  TelescopeSpec(ex.MID_B1_Rebase), # — hybrid；z=[0.25, 3.25]；自带 n(x) 缺失但 hybrid 不读
    "mid_b1_octave_hybrid":  TelescopeSpec(ex.MID_B1_Octave), # — hybrid；z=[0.39, 2.03]
    "mid_b2_rebase_hybrid":  TelescopeSpec(ex.MID_B2_Rebase), # — hybrid；z=[-0.10, 0.77]
    "mid_b2_octave_hybrid":  TelescopeSpec(ex.MID_B2_Octave), # — hybrid；z=[-0.10, 0.98]
    "mid_b1_base_dish":      TelescopeSpec(ex.MID_B1_Base),   # ⚠️ 缺 nx_SKAMREF2_dec30_254.dat（dish 不读）；z=[0.25, 3.25]
    "mid_b1_alt_dish":       TelescopeSpec(ex.MID_B1_Alt),    # ⚠️ 缺 nx_SKAMREF2_dec30_200.dat（dish 不读）；z=[0.67, 2.49]
    "mid_b2_base_dish":      TelescopeSpec(ex.MID_B2_Base),   # z=[-0.10, 0.77]
    "mid_b2_upd_dish":       TelescopeSpec(ex.MID_B2_Upd),    # z=[-0.10, 0.77]
    "mid_b2_alt_dish":       TelescopeSpec(ex.MID_B2_Alt),    # z=[-0.10, 1.00]
    "mid_b2_alt2_dish":      TelescopeSpec(ex.MID_B2_Alt2),   # z=[-0.01, 1.00]
    "mid_b2_base_interferom": TelescopeSpec(ex.MID_B2_Base,  mode_override="interferom"),
    "mid_b2_upd_interferom":  TelescopeSpec(ex.MID_B2_Upd,   mode_override="interferom"),
    "mid_b2_alt_interferom":  TelescopeSpec(ex.MID_B2_Alt,   mode_override="interferom"),
    "mid_b2_alt2_interferom": TelescopeSpec(ex.MID_B2_Alt2,  mode_override="interferom"),
    #   ✅ nx_SKAMREF2_dec30.dat

    # ---------------- MeerKAT（配置自带 dish，干涉需 override）----------------
    "meerkat_b1_interferom": TelescopeSpec(ex.MeerKATb1, mode_override="interferom"),
    #   ✅ nx_MKREF2_dec30.dat；红移 z=[0.30, 1.64]
    "meerkat_b2_interferom": TelescopeSpec(ex.MeerKATb2, mode_override="interferom"),
    #   ✅ nx_MKREF2_dec30.dat；红移 z=[-0.10, 0.77]

    # ---------------- KAT7（同上）----------------
    "kat7_interferom":       TelescopeSpec(ex.KAT7, mode_override="interferom"),
    #   ✅ nx_KAT7_dec30.dat；红移 z=[-0.10, 0.38]

    # ---------------- ASKAP（同上）----------------
    "askap_interferom":      TelescopeSpec(ex.ASKAP, mode_override="interferom"),
    #   ✅ nx_ASKAP_dec30.dat；红移 z=[0.32, 1.22]

    # ---------------- JVLA（同上）----------------
    "jvla_interferom":       TelescopeSpec(ex.JVLA, mode_override="interferom"),
    #   ✅ nx_VLAD_dec90.dat；红移 z=[-0.10, 0.61]

    # ---------------- FAST（单碟）----------------
    "fast_dish":             TelescopeSpec(ex.FAST),          # 500m 单碟；红移 z=[0.32, 2.74]

    # ---------------- GBT / GBTHIM（单碟）----------------
    "gbt_dish":              TelescopeSpec(ex.GBT),           # 100m 单碟；红移 z=[0.44, 1.28]
    "gbthim_dish":           TelescopeSpec(ex.GBTHIM),        # 100m 单碟+7 波束；红移 z=[0.48, 1.22]

    # ---------------- 其他单碟望远镜 ----------------
    "parkes_dish":           TelescopeSpec(ex.Parkes),        # 红移 z=[-0.10, 0.42]
    "gmrt_dish":             TelescopeSpec(ex.GMRT),          # 红移 z=[-0.10, 0.61]
    "wsrt_dish":             TelescopeSpec(ex.WSRT),          # 红移 z=[0.08, 1.22]
    "apertif_dish":          TelescopeSpec(ex.APERTIF),       # 红移 z=[-0.01, 0.61]
    "vlba_dish":             TelescopeSpec(ex.VLBA),          # 红移 z=[-0.10, 0.38]
    "bingo_dish":            TelescopeSpec(ex.BINGO),         # 红移 z=[0.02, 0.67]

    # ---------------- 早期干涉阵列（uniform，自带 Dmin/Dmax）----------------
    "baobab32_interferom":   TelescopeSpec(ex.BAOBAB32),      # 🔢 uniform；红移 z=[0.48, 1.56]
    "baobab128_interferom":  TelescopeSpec(ex.BAOBAB128),     # 🔢 uniform；红移 z=[0.48, 1.56]
    "kzn_interferom":        TelescopeSpec(ex.KZN),           # 🔢 uniform；红移 z=[0.48, 1.56]
    "mfaa_interferom":       TelescopeSpec(ex.MFAA),          # 🔢 uniform；红移 z=[0.39, 2.35]
    "aera3_interferom":      TelescopeSpec(ex.AERA3),         # 🔢 uniform；红移 z=[0.08, 1.22]

    # ---------------- 示例实验（论文 illustrative）----------------
    "exptS_dish":            TelescopeSpec(ex.exptS),         # 单碟；红移 z=[0.19, 0.97]
    "exptCV_dish":           TelescopeSpec(ex.exptCV),        # CV 极限单碟；红移 z=[0.55, 2.28]
    "exptM_interferom":      TelescopeSpec(ex.exptM),         # 🔢 uniform（Dmin=4/Dmax=53）；z=[0.32, 1.56]
}


# ============================================================================
# ============================================================================
# 2b. mock 参数与目标望远镜（解析 + mock(z 对齐)）
#     mock z 覆盖 0.5–3.0（6 层、Δz=0.5）；k 越界 → f_arr=0 退回纯解析
# ============================================================================
from dataclasses import replace as _dc_replace

MOCK_FILE      = "/home/liuhk/research/mock_fisher_arr.joblib"
MOCK_Z_LO      = 0.5
MOCK_Z_HI      = 3.0
MOCK_Z_CENTERS = np.linspace(MOCK_Z_LO, MOCK_Z_HI, 6)   # mock 6 层 z 中心
_MOCKED_KEYS   = ["ska_mid1_interferom", "chime_icyl", "hirax_interferom",
                  "tianlai_icyl", "cv_z0to3_interferom", "meerkat_b1_interferom"]

for _k in _MOCKED_KEYS:
    if _k in TELESCOPES:
        TELESCOPES[_k] = _dc_replace(TELESCOPES[_k], mock_file=MOCK_FILE)

# 2. 通用小工具
# ============================================================================
def fisher_submatrix(F: np.ndarray, names: List[str], keep: List[str]) -> np.ndarray:
    idx = [names.index(k) for k in keep if k in names]
    if len(idx) != len(keep):
        missing = [k for k in keep if k not in names]
        raise ValueError(f"Requested params not in Fisher names: {missing}. Available: {names}")
    idx = np.array(idx, dtype=int)
    return F[np.ix_(idx, idx)]


def expt_to_f_sky(expt: dict) -> float:
    return float(expt["Sarea"]) / (4.0 * np.pi)


def expt_to_zmin_zmax(expt: dict) -> Tuple[float, float]:
    nu_line = float(expt["nu_line"])
    numax = float(expt["survey_numax"])
    dnutot = float(expt["survey_dnutot"])
    zmin = nu_line / numax - 1.1028
    zmax = nu_line / (numax - dnutot) - 0.8083
    return float(zmin), float(zmax)


def bin_dnutot_MHz(nu_line_MHz: float, z_low: float, z_high: float) -> float:
    # Δν for bin [z_low, z_high]: nu(z)=nu_line/(1+z)
    nu_high = nu_line_MHz / (1.0 + z_low)   # 高频率对应低 z
    nu_low = nu_line_MHz / (1.0 + z_high)
    return max(float(nu_high - nu_low), 1e-12)


# ============================================================================
# 3. SurveyConfig：望远镜装配（唯一配置入口，复刻 baofisher 的 mode/n(x) 规则）
# ============================================================================
@dataclass
class SurveyConfig:
    telescope: str = "chime"
    t_obs_hours: float = 100000.0           # 观测时间（小时），唯一时间入口
    delta_z: float = 0.5
    k_min: float = 5.0e-3 
    k_max: float = 0.535
    nk: int = 50
    n_mu: int = 100
    include_Cfg: bool = False               # 是否在噪声里加前景项 Cfg
    kref_T_h_Mpc: float = 2e-3              # 转移函数归一化尺度 (h/Mpc)
    # 由 load() 自动填充
    f_sky: float = 0.0
    t_obs_total: float = 0.0
    z_bins: Tuple[float, float] = (0.0, 0.0)
    expt: dict = field(default_factory=dict, repr=False)

    def load(self) -> dict:
        if self.telescope not in TELESCOPES:
            raise KeyError(f"未知望远镜 {self.telescope!r}，可用: {sorted(TELESCOPES)}")
        spec = TELESCOPES[self.telescope]
        ex = copy.deepcopy(spec.expt)

        # 规则1：mode = override 或配置自带（'cylinder'→'icyl' 自动翻译）
        mode = normalize_mode(spec.mode_override if spec.mode_override is not None else ex.get("mode", ""))
        ex["mode"] = mode
        ex["ttot"] = self.t_obs_hours * 3600 * 1e6     # baofisher 单位: sec * MHz

        # 规则1b：extra 始终应用（文件型与 uniform 都补齐 Ncyl/cyl_area/Dmin/Dmax 等）
        ex.update(spec.extra)

        # 规则2：n(x) 由 mode 驱动自动加载（同 baofisher.fisher()：干涉且有 n(x) 字符串才加载）
        if mode[0] == "i" and isinstance(ex.get("n(x)"), str):
            p = Path(BAO21CM_ROOT) / ex["n(x)"]
            if p.exists():
                ex["n(x)"] = bf.load_interferom_file(str(p))
            else:
                warnings.warn(f"[{self.telescope}] n(x) 文件缺失 {p} → 降级 uniform")
                ex.pop("n(x)", None)

        # 规则3：uniform 分支（extra 已由规则1b应用；仅校验 Dmin/Dmax）
        if not callable(ex.get("n(x)")):
            ex.pop("n(x)", None)
            if mode[0] == "i" and not ({"Dmin", "Dmax"} <= ex.keys()):
                raise ValueError(f"[{self.telescope}] 干涉分支需补 Dmin/Dmax（加到 extra）")

        self.expt = ex
        self.f_sky = expt_to_f_sky(ex)
        self.t_obs_total = self.t_obs_hours * 3600
        self.z_bins = expt_to_zmin_zmax(ex)
        return ex


# ============================================================================
# 4. CosmologyEngine：信号模型 + baofisher 噪声
# ============================================================================
class CosmologyEngine:
    def __init__(
        self,
        config: SurveyConfig,
        # ------ Planck 2018 baseline 参数 ------
        lnA=np.log(2.1e-9),
        Omega_m=0.3153,
        omega_b=0.02237,
        omega_cdm=0.1200,
        n_s=0.9649,
        # ---------------------------------------
        w0=-1.0,
        sum_mnu=0.06,
        f_NL=0.0,
    ):
        self.config = config
        self.lnA = float(lnA)
        self.Omega_m = float(Omega_m)
        self.omega_b = float(omega_b)
        self.omega_cdm = float(omega_cdm)
        self.n_s = float(n_s)
        self.w0 = float(w0)
        self.sum_mnu = float(sum_mnu)
        self.f_NL = float(f_NL)

        self.cosmo = Class()
        self.class_params = {
            "output": "mPk",
            "P_k_max_h/Mpc": 5.0,
            "non linear": "halofit",
            "z_max_pk": float(self.config.z_bins[1] + 0.5),
        }
        self.transfer_fn_bf = None
        self.update_cosmology()

    def update_cosmology(self):
        A_s = float(np.exp(self.lnA))
        h = float(np.sqrt((self.omega_b + self.omega_cdm) / self.Omega_m))
        self.h = h
        self.class_params.update(
            {
                "h": h,
                "omega_b": self.omega_b,
                "omega_cdm": self.omega_cdm,
                "Omega_Lambda": 1.0 - self.Omega_m,
                "A_s": A_s,
                "n_s": self.n_s,
                "w0_fld": self.w0,
                "m_ncdm": str(self.sum_mnu),
                "N_ncdm": "1",
            }
        )
        self.cosmo.set(self.class_params)
        self.cosmo.compute()

    def get_baofisher_transfer(self):
        """baofisher 的 iscalefn(k) = 1 / (T(k) * k^2)。"""
        if self.transfer_fn_bf is not None:
            return self.transfer_fn_bf
        cosmo_bf = {
            "h": self.h,
            "omega_M_0": self.Omega_m,
            "omega_b_0": self.omega_b / (self.h**2),
            "omega_lambda_0": 1.0 - self.Omega_m,
            "ns": self.n_s,
            "w0": self.w0,
            "wa": 0.0,
            "mnu": self.sum_mnu,
            "sigma_8": 0.8,
            "fNL": 0.0,
        }
        cachefile = "transfer_bf_cache.dat"
        kref_phys = float(self.config.kref_T_h_Mpc) * self.h
        iscalefn, _, _ = bf.deriv_transfer(
            cosmo_bf, cachefile, kmax=bf.CAMB_KMAX, kref=kref_phys, force=True
        )
        self.transfer_fn_bf = iscalefn
        return iscalefn

    def growth_D(self, z: float) -> float:
        return float(self.cosmo.scale_independent_growth_factor(float(z)))

    def average_brightness_temperature_mK(self, z: float) -> float:
        return float(5.5919e-02 + 2.3242e-01 * z - 2.4136e-02 * z**2)

    def bias(self, z: float) -> float:
        bHI0 = float(ex.cosmo.get("bHI0", 0.677))
        return float(bf.bias_HI(float(z), {"bHI0": bHI0}))

    def scale_dependent_bias(self, k_Mpc: np.ndarray, z: float) -> np.ndarray:
        k = np.atleast_1d(k_Mpc).astype(float)
        b0 = self.bias(float(z))
        alpha = self.alpha_local_png(k, float(z))
        return 3.0 * (b0 - 1.0) * self.f_NL * alpha

    def HI_bias(self, k_Mpc: np.ndarray, z: float) -> np.ndarray:
        k = np.atleast_1d(k_Mpc).astype(float)
        return self.bias(float(z)) + self.scale_dependent_bias(k, float(z))

    def growth_rate(self, z: float) -> float:
        return float(self.cosmo.scale_independent_growth_factor_f(float(z)))

    def RSD_beta(self, k_Mpc: np.ndarray, z: float) -> np.ndarray:
        return self.growth_rate(float(z)) / self.HI_bias(k_Mpc, float(z))

    def power_spectrum(self, k_Mpc: np.ndarray, z: float) -> np.ndarray:
        k_arr = np.atleast_1d(k_Mpc).astype(float)
        return np.array([self.cosmo.pk(ki, float(z)) for ki in k_arr])

    def comoving_distance_Mpc(self, z: float) -> float:
        return float(self.cosmo.comoving_distance(float(z)))

    def alpha_local_png(self, k_Mpc: np.ndarray, z: float) -> np.ndarray:
        k = np.atleast_1d(k_Mpc).astype(float)
        iscalefn = self.get_baofisher_transfer()
        inv_T_k2 = iscalefn(k)
        D = max(float(self.growth_D(float(z))), 1e-30)
        H0 = 100.0 * self.h
        pref = (H0 / C_LIGHT_KMS) ** 2 * self.Omega_m * DELTA_C
        return pref * inv_T_k2 * (1.0 / D)

    def HI_power_spectrum(self, k_Mpc: np.ndarray, z: float, mu: np.ndarray) -> np.ndarray:
        k = np.atleast_1d(k_Mpc).astype(float)
        mu = np.atleast_1d(mu).astype(float)
        Tbar = self.average_brightness_temperature_mK(float(z))
        b = self.HI_bias(k, float(z))
        beta = self.RSD_beta(k, float(z))
        Pm = self.power_spectrum(k, float(z))
        mu2 = mu[None, :] ** 2
        sigma_nl = getattr(self, "sigma_nl", 7.0)
        damping = np.exp(-mu2 * (k[:, None] * sigma_nl) ** 2)
        return (
            (Tbar**2)
            * (b[:, None] ** 2)
            * (Pm[:, None] * (1.0 + beta[:, None] * mu2) ** 2)
            * damping
        )

    # ---- 单点噪声（供绘图用） ----
    def P_noise(
        self,
        k_Mpc: float,
        mu: float,
        z_mid: float,
        z_low: float,
        z_high: float,
        expt: dict,
        include_Cfg: bool = True,
    ) -> float:
        k_Mpc = float(k_Mpc)
        mu = float(mu)
        z_mid = float(z_mid)

        r_Mpc = float(self.cosmo.comoving_distance(z_mid))
        H_class = float(self.cosmo.Hubble(z_mid))
        rnu_Mpc = (1.0 + z_mid) ** 2 / H_class

        q = r_Mpc * k_Mpc * np.sqrt(max(0.0, 1.0 - mu**2))
        y = rnu_Mpc * k_Mpc * abs(mu)

        cosmo_dict = {
            "z": z_mid, "aperp": 1.0, "apar": 1.0,
            "r": r_Mpc, "rnu": rnu_Mpc, "ns": float(self.n_s),
        }
        expt_dict = copy.deepcopy(expt)
        expt_dict["dnutot"] = bin_dnutot_MHz(float(expt_dict["nu_line"]), float(z_low), float(z_high))

        _buf = io.StringIO()
        with contextlib.redirect_stdout(_buf):
            cn = bf.Cnoise(np.array([q]), np.array([y]), cosmo_dict, expt_dict, cv=False)[0]
            cf = (
                bf.Cfg(np.array([q]), np.array([y]), cosmo_dict, expt_dict)[0]
                if include_Cfg else 0.0
            )

        if (not np.isfinite(cn)) or (cn >= 0.5 * BF_INF_NOISE):
            return float("inf")
        if include_Cfg and ((not np.isfinite(cf)) or (cf >= 0.5 * BF_INF_NOISE)):
            return float("inf")
        return float((cn + cf) * (r_Mpc**2) * rnu_Mpc)

    # ---- 向量化噪声（compute_bin_data 用，替代 5000 次单点循环） ----
    def P_noise_grid(
        self,
        ks: np.ndarray,
        mu: np.ndarray,
        z_mid: float,
        z_low: float,
        z_high: float,
        expt: dict,
        include_Cfg: bool = True,
    ) -> np.ndarray:
        k2 = np.atleast_2d(np.asarray(ks, dtype=float)).T        # (nk,1)
        m2 = np.atleast_2d(np.asarray(mu, dtype=float))          # (1,n_mu)
        r = float(self.cosmo.comoving_distance(z_mid))
        H = float(self.cosmo.Hubble(z_mid))
        rnu = (1.0 + z_mid) ** 2 / H

        q = r * k2 * np.sqrt(np.clip(1.0 - m2**2, 0.0, None))
        y = rnu * k2 * np.abs(m2)

        cosmo_dict = {
            "z": z_mid, "aperp": 1.0, "apar": 1.0,
            "r": r, "rnu": rnu, "ns": float(self.n_s),
        }
        ex = copy.deepcopy(expt)
        ex["dnutot"] = bin_dnutot_MHz(float(ex["nu_line"]), float(z_low), float(z_high))

        _buf = io.StringIO()
        with contextlib.redirect_stdout(_buf):
            cn = bf.Cnoise(q, y, cosmo_dict, ex, cv=False)
            cf = bf.Cfg(q, y, cosmo_dict, ex) if include_Cfg else 0.0

        bad = ~np.isfinite(cn) | (cn >= 0.5 * BF_INF_NOISE)
        if include_Cfg:
            bad |= ~np.isfinite(cf) | (cf >= 0.5 * BF_INF_NOISE)
        return np.where(bad, np.inf, (cn + cf) * r**2 * rnu)


# ============================================================================
# 5. HIFisherForecast：Fisher 矩阵
# ============================================================================
class HIFisherForecast:
    def __init__(self, engine: CosmologyEngine, config: SurveyConfig):
        self.cosmo = engine
        self.config = config
        # 约束全部 7 个 Planck 参数：f_NL 用解析偏导，其余 6 个用数值偏导
        self.parameters = ["lnA", "Omega_m", "omega_b", "omega_cdm", "n_s", "w0", "f_NL"]
        self.eps = {}
        for p in self.parameters:
            if p == "f_NL":
                self.eps[p] = 0.1
            else:
                v = float(getattr(self.cosmo, p))
                self.eps[p] = 0.025 * abs(v) if abs(v) > 0 else 1e-6

    def volume_shell(self, z1: float, z2: float) -> float:
        r1 = self.cosmo.comoving_distance_Mpc(z1)
        r2 = self.cosmo.comoving_distance_Mpc(z2)
        return 4.0 * np.pi / 3.0 * (r2**3 - r1**3) * float(self.config.f_sky)

    def _dP_HI_dfNL_analytic(self, k_arr, z, mu_arr):
        k = np.atleast_1d(k_arr).astype(float)
        mu = np.atleast_1d(mu_arr).astype(float)
        Tbar = self.cosmo.average_brightness_temperature_mK(z)
        b0 = self.cosmo.bias(z)
        alpha = self.cosmo.alpha_local_png(k, z)
        db_dfNL = 3.0 * (b0 - 1.0) * alpha
        b_tot = self.cosmo.HI_bias(k, z)
        beta = self.cosmo.RSD_beta(k, z)
        Pm = self.cosmo.power_spectrum(k, z)
        mu2 = mu[None, :] ** 2
        sigma_nl = getattr(self.cosmo, "sigma_nl", 7.0)
        damping = np.exp(-mu2 * (k[:, None] * sigma_nl) ** 2)
        return (
            (Tbar**2)
            * 2.0 * b_tot[:, None] * db_dfNL[:, None]
            * Pm[:, None] * (1.0 + beta[:, None] * mu2)
            * damping
        )

    def dP_HI_dparam(self, k_arr, z, mu_arr, param: str) -> np.ndarray:
        if param == "f_NL":
            return self._dP_HI_dfNL_analytic(k_arr, z, mu_arr)
        orig = float(getattr(self.cosmo, param))
        dp = []
        for sgn in (+1, -1):
            setattr(self.cosmo, param, orig + sgn * self.eps[param])
            if param != "f_NL":
                self.cosmo.update_cosmology()
            P = self.cosmo.HI_power_spectrum(k_arr, z, mu_arr)
            dp.append(P)
        setattr(self.cosmo, param, orig)
        if param != "f_NL":
            self.cosmo.update_cosmology()
        return (dp[0] - dp[1]) / (2.0 * self.eps[param])

    def compute_bin_data(self, expt: dict, include_Cfg: bool = True) -> Tuple:
        """计算 P_HI / P_noise / derivs（解析与 mock 共用，与 mock 数据无关）。"""
        zmin, zmax = self.config.z_bins
        dz = float(self.config.delta_z)
        n_bins = int(round((zmax - zmin) / dz))
        z_edges = np.linspace(zmin, zmax, n_bins + 1)
        if z_edges[0] != zmin:
            z_edges = np.concatenate(([zmin], z_edges))
        if z_edges[-1] < zmax:
            z_edges = np.concatenate((z_edges, [zmax]))
        else:
            z_edges[-1] = zmax

        ks = np.logspace(np.log10(self.config.k_min), np.log10(self.config.k_max), int(self.config.nk))
        mu = np.linspace(-1.0, 1.0, int(self.config.n_mu))
        npar = len(self.parameters)
        bin_data = []

        for i in range(len(z_edges) - 1):
            z_low, z_high = float(z_edges[i]), float(z_edges[i + 1])
            z_mid = 0.5 * (z_low + z_high)

            P_HI = self.cosmo.HI_power_spectrum(ks, z_mid, mu)
            P_noise = self.cosmo.P_noise_grid(ks, mu, z_mid, z_low, z_high, expt, include_Cfg)

            derivs = np.zeros((npar, ks.size, mu.size))
            for p_idx, p in enumerate(self.parameters):
                derivs[p_idx] = self.dP_HI_dparam(ks, z_mid, mu, p)

            bin_data.append(
                {
                    "z_low": z_low, "z_high": z_high, "z_mid": z_mid,
                    "r_Mpc": float(self.cosmo.comoving_distance_Mpc(z_mid)),
                    "H_z": float(self.cosmo.cosmo.Hubble(z_mid)),
                    "k": ks.copy(), "mu": mu.copy(),
                    "P_HI": P_HI.copy(), "P_noise": P_noise.copy(),
                    "derivs": {p: d.copy() for p, d in zip(self.parameters, derivs)},
                }
            )
        return z_edges, bin_data

    def fisher_sum_analytic(self, z_edges_and_bin_data: Tuple) -> Tuple[np.ndarray, np.ndarray, List[dict]]:
        """解析方差 (P_HI+P_noise)^2 求和（同 4.30.1 原始方法）。"""
        z_edges, bin_data = z_edges_and_bin_data
        mu = bin_data[0]["mu"]
        mu_w = np.ones_like(mu) * (2.0 / mu.size)
        npar = len(self.parameters)
        F = np.zeros((npar, npar))

        for bd in bin_data:
            derivs = np.array([bd["derivs"][p] for p in self.parameters])
            ks = bd["k"]
            dlnk = np.log(ks[1] / ks[0])
            V = self.volume_shell(bd["z_low"], bd["z_high"])
            weight_k = V * ks**3 / (8.0 * np.pi**2) * dlnk
            tot = bd["P_HI"] + bd["P_noise"]
            bad = ~np.isfinite(tot) | (tot <= 0.0)
            tot = tot.copy()
            tot[bad] = np.inf
            with np.errstate(over="ignore", invalid="ignore"):
                inv_tot2 = 1.0 / (tot * tot)
            for a in range(npar):
                for b in range(npar):
                    integ_mu = np.sum(mu_w[None, :] * (derivs[a] * derivs[b]) * inv_tot2, axis=1)
                    F[a, b] += np.sum(integ_mu * weight_k)
        return F, z_edges, bin_data

    def fisher_sum_mock_zinterp(self, z_edges_and_bin_data, mock_list):
        """解析方差 + mock 方差按红移 z 对齐后求和。

        Var_tot = (P_HI+P_noise)^2 + f_arr(z_mid; k⊥,k∥)·P_HI^2
        - z 维：解析 bin 的 z_mid 在 mock z 中心之间线性插值（越界 clip 到边界）
        - k 维：越出 mock 网格 → f_arr=0 → 该项退回纯解析方差
        返回与 fisher_sum_analytic 相同的 (F, z_edges, bin_data)。
        """
        from scipy.interpolate import RegularGridInterpolator

        z_edges, bin_data = z_edges_and_bin_data
        ks = bin_data[0]["k"]
        mu = bin_data[0]["mu"]
        z_mock = np.asarray(MOCK_Z_CENTERS, dtype=float)
        assert z_mock.size == len(mock_list), "mock z 中心数 != mock 层数"

        # 各层共用同一 (k⊥,k∥) 网格 → 取第 0 层坐标，值沿 z 堆成 3D
        kper0, kpar0 = mock_list[0]["kper_bin"], mock_list[0]["kpar_bin"]
        cube = np.stack([m["fisher_arr"] for m in mock_list], axis=0)   # (nz,200,200)
        for m in mock_list[1:]:
            assert np.array_equal(m["kper_bin"], kper0) and np.array_equal(m["kpar_bin"], kpar0), \
                "mock 各层 k 网格不一致，无法堆成三维"
        interp3d = RegularGridInterpolator((z_mock, kper0, kpar0), cube,
                                           bounds_error=False, fill_value=0.0)

        z_mids = np.array([bd["z_mid"] for bd in bin_data], dtype=float)
        z_clip = np.clip(z_mids, z_mock[0], z_mock[-1])

        # (k,μ) → (k⊥,k∥)（与 A3 fisher_sum_mock 同一映射）
        kper = ks[:, None] * np.sqrt(np.clip(1.0 - mu[None, :] ** 2, 0.0, None))
        kpar = ks[:, None] * np.abs(mu[None, :])

        dlnk = np.log(ks[1] / ks[0])
        mu_w = np.ones_like(mu) * (2.0 / mu.size)
        npar = len(self.parameters)
        F = np.zeros((npar, npar))

        for i, bd in enumerate(bin_data):
            npts = ks.size * mu.size
            pts = np.stack([
                np.full(npts, z_clip[i]),
                np.broadcast_to(kper, (ks.size, mu.size)).ravel(),
                np.broadcast_to(kpar, (ks.size, mu.size)).ravel(),
            ], axis=-1)
            f_arr = interp3d(pts).reshape(ks.size, mu.size)             # 越界=0
            Var_mock = f_arr * (bd["P_HI"] ** 2)                        # 物理 mock 方差

            V = self.volume_shell(bd["z_low"], bd["z_high"])
            tot = bd["P_HI"] + bd["P_noise"]
            bad = ~np.isfinite(tot) | (tot <= 0.0)
            tot_safe = np.where(bad, np.inf, tot)
            weight_k = V * ks ** 3 / (8.0 * np.pi ** 2) * dlnk
            geometric_weight = weight_k[:, None] * mu_w[None, :]

            with np.errstate(over="ignore", invalid="ignore"):
                combined_inv_var = 1.0 / (tot_safe ** 2 + Var_mock) * geometric_weight

            derivs = np.array([bd["derivs"][p] for p in self.parameters])
            for a in range(npar):
                for b in range(npar):
                    F[a, b] += np.sum(derivs[a] * derivs[b] * combined_inv_var)
        return F, z_edges, bin_data



# ============================================================================
# 6. 输出层：约束汇总表 + k 空间曲线图（统一口径：全部来自 bin_data / F_*）
# ============================================================================
OUT_DIR = Path("outputs")

# ====================== y 轴截距设计（物理掩码 + 鲁棒百分位窗口） ======================
Y_Q_LO = 2.0             # 有效区间下分位（%）
Y_Q_HI = 98.0            # 有效区间上分位（%）
Y_MARGIN_DEC = 0.2       # 窗口外扩（log10 量级）
NOISE_MASK_FACTOR = 1e4  # P_noise ≥ NOISE_MASK_FACTOR×max(P_HI) → 无效（前景楔/无基线区）


def robust_ylim(values_list: List[np.ndarray], q_lo: float = Y_Q_LO, q_hi: float = Y_Q_HI,
                margin_dec: float = Y_MARGIN_DEC, symlog: bool = False) -> Optional[Tuple[float, float]]:
    """从多组曲线的有限值求鲁棒 y 轴窗口 [ylo, yhi]。

    - log 图：取有限正值在 log10 空间的 [q_lo, q_hi] 分位，外扩 margin_dec 量级。
    - symlog 图：取带符号有限值的 [q_lo, q_hi] 分位，外扩相对边距。
    - 有效点不足或分布退化时返回 None（交回 matplotlib 自动缩放）。
    """
    if not values_list:
        return None
    if symlog:
        vals = np.concatenate([np.asarray(v, dtype=float)[np.isfinite(v)] for v in values_list])
        if vals.size < 4:
            return None
        lo, hi = float(np.percentile(vals, q_lo)), float(np.percentile(vals, q_hi))
        if not (np.isfinite(lo) and np.isfinite(hi)) or hi <= lo:
            return None
        pad = max(1e-12, 0.05 * (hi - lo))
        return (lo - pad, hi + pad)
    logs = np.concatenate([
        np.log10(np.asarray(v, dtype=float)[np.isfinite(v) & (v > 0)]) for v in values_list
    ])
    if logs.size < 4:
        return None
    lo, hi = float(np.percentile(logs, q_lo)), float(np.percentile(logs, q_hi))
    if not (np.isfinite(lo) and np.isfinite(hi)) or hi <= lo:
        return None
    pad = max(margin_dec, 0.05 * (hi - lo))
    return (10.0 ** (lo - pad), 10.0 ** (hi + pad))


def _ref_z_idx(result: dict) -> int:
    """参考 z-bin：取中间 bin。"""
    return (len(result["bin_data"]) - 1) // 2


def _mu_idx(result: dict, mu_ref: float = 0.5) -> int:
    return int(np.argmin(np.abs(result["bin_data"][0]["mu"] - mu_ref)))


def build_curves(result: dict, mu_ref: float = 0.5) -> dict:
    """组装参考 z 处的 k 空间曲线：P_HI / P_noise / P_tot / SNR / derivs。

    P_noise 应用物理掩码：楔形/无基线区返回的 ~10²⁰⁰ 巨大有限值视为无效（nan），
    P_tot/SNR 由掩码后噪声派生。只影响绘图，不影响 bin_data 与 Fisher。
    """
    zi = _ref_z_idx(result)
    imu = _mu_idx(result, mu_ref)
    bd = result["bin_data"][zi]
    ks = bd["k"]
    P_HI = np.asarray(bd["P_HI"][:, imu])
    P_noise_raw = np.asarray(bd["P_noise"][:, imu])
    P_HI_max = float(np.max(P_HI)) if P_HI.size else 1.0
    bad = ~np.isfinite(P_noise_raw) | (P_noise_raw >= NOISE_MASK_FACTOR * P_HI_max)
    P_noise = np.where(bad, np.nan, P_noise_raw)
    with np.errstate(divide="ignore", invalid="ignore"):
        P_tot = P_HI + P_noise
        SNR = P_HI / P_noise
    curves = {
        "k": ks,
        "z_ref": float(bd["z_mid"]),
        "mu": float(bd["mu"][imu]),
        "P_HI": P_HI, "P_noise": P_noise, "P_tot": P_tot, "SNR": SNR,
        "derivs": {p: np.asarray(bd["derivs"][p][:, imu]) for p in result["parameters"]},
    }
    result["curves"] = curves
    return curves


def _interp_mock_fisher_arr(mock_list, z_ref, ks, mu):
    """把 mock 的 fisher_arr 按 (z_ref, k⊥, k∥) 三维插值到 (ks, mu)（与求和同规则）。"""
    from scipy.interpolate import RegularGridInterpolator
    z_mock = np.asarray(MOCK_Z_CENTERS, dtype=float)
    kper0, kpar0 = mock_list[0]["kper_bin"], mock_list[0]["kpar_bin"]
    cube = np.stack([m["fisher_arr"] for m in mock_list], axis=0)
    interp = RegularGridInterpolator((z_mock, kper0, kpar0), cube,
                                     bounds_error=False, fill_value=0.0)
    zz = float(np.clip(z_ref, z_mock[0], z_mock[-1]))
    kper = ks[:, None] * np.sqrt(np.clip(1.0 - mu[None, :] ** 2, 0.0, None))
    kpar = ks[:, None] * np.abs(mu[None, :])
    npts = ks.size * mu.size
    pts = np.stack([np.full(npts, zz),
                    np.broadcast_to(kper, (ks.size, mu.size)).ravel(),
                    np.broadcast_to(kpar, (ks.size, mu.size)).ravel()], axis=-1)
    return interp(pts).reshape(ks.size, mu.size)


def build_mock_curves(forecast: dict, mu_ref: float = 0.5) -> Optional[dict]:
    """组装参考 z 处 mock 方差曲线（z 对齐；越界 f_arr=0 → nan）。"""
    ana = forecast["analytic"]
    if "mock" not in forecast:
        return None
    zi = _ref_z_idx(ana)
    imu = _mu_idx(ana, mu_ref)
    bd = ana["bin_data"][zi]
    ks = bd["k"]
    mu = bd["mu"]
    mock_list = forecast["mock_list"]
    farr = _interp_mock_fisher_arr(mock_list, float(bd["z_mid"]), ks, mu)
    P_HI = np.asarray(bd["P_HI"])
    P_noise_raw = np.asarray(bd["P_noise"])
    P_HI_max = float(np.max(P_HI)) if P_HI.size else 1.0
    bad_noise = ~np.isfinite(P_noise_raw) | (P_noise_raw >= NOISE_MASK_FACTOR * P_HI_max)
    P_noise = np.where(bad_noise, np.nan, P_noise_raw)
    Var_mock = np.where(np.isfinite(farr) & (farr > 0), farr * P_HI ** 2, np.nan)
    with np.errstate(over="ignore", invalid="ignore"):
        Var_tot_mock = (P_HI + P_noise) ** 2 + Var_mock
    return {
        "k": ks, "z_ref": float(bd["z_mid"]), "mu": float(mu[imu]),
        "Var_mock": Var_mock[:, imu],
        "Var_tot_mock": Var_tot_mock[:, imu],
    }


def _curve_value(curves: dict, key: str) -> np.ndarray:
    """支持 'P_HI' 或 'derivs.f_NL' 形式访问曲线。"""
    if key.startswith("derivs."):
        return curves["derivs"][key.split(".", 1)[1]]
    return curves[key]


def _plot_curves(curve_list: List[Tuple[str, dict]], key: str, ylabel: str, fname: str,
                 *, symlog_y: bool = False) -> None:
    """通用多曲线绘图：curve_list = [(label, curves_dict), ...]。

    y 轴用 robust_ylim（[p2,p98] + 外扩）聚焦有效区间；标题标注裁剪点数。
    """
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(9, 6))
    linestyles = ["-", "--", "-.", ":"]
    values = []
    for i, (label, c) in enumerate(curve_list):
        k = np.asarray(c["k"], dtype=float)
        y = np.asarray(_curve_value(c, key), dtype=float)
        if symlog_y:
            y = np.where(np.isfinite(y), y, np.nan)
        else:
            y = np.where(np.isfinite(y) & (y > 0), y, np.nan)
        values.append(y[np.isfinite(y)])
        if np.all(~np.isfinite(y)):
            warnings.warn(f"[plot {Path(fname).stem}] {label}: 无有效点（全被掩码）")
        ax.plot(k, y, linestyle=linestyles[i % len(linestyles)], linewidth=2,
                label=f"{label}  (z={c.get('z_ref', float('nan')):.2f})")
    ax.set_xscale("log")
    if symlog_y:
        ax.set_yscale("symlog", linthresh=1e-12)
    else:
        ax.set_yscale("log")
    ax.set_xlabel(r"$k$ [Mpc$^{-1}$]")
    ax.set_ylabel(ylabel)
    ylim = robust_ylim(values, symlog=symlog_y)
    if ylim is not None:
        lo, hi = ylim
        allv = np.concatenate(values) if values else np.array([])
        nclip = int(np.sum((allv < lo) | (allv > hi))) if allv.size else 0
        ax.set_ylim(lo, hi)
        ax.set_title(f"{Path(fname).stem}  (y-window: {nclip} pts clipped)")
    else:
        ax.set_title(Path(fname).stem)
    ax.grid(True, which="both", ls="--", alpha=0.4)
    ax.legend(fontsize=9, ncol=1)
    fig.tight_layout()
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    fig.savefig(OUT_DIR / fname, dpi=150)
    plt.show()
    plt.close(fig)


# ---- ② 方差图：signal/noise/analytic 用 realspace 方差数据（var vs z，来自 variance_table）----
def _plot_variance_vs_z(results: List[dict], key: str, ylabel: str, fname: str) -> None:
    """单个方差量随 z 变化：多望远镜同图（semilogy + 鲁棒窗口）。"""
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(9, 6))
    linestyles = ["-", "--", "-.", ":"]
    markers = ["o", "s", "^", "d"]
    values = []
    for i, res in enumerate(results):
        rows = res["variance_table"]
        z = np.asarray([r["z_mid"] for r in rows], dtype=float)
        var = np.asarray([r[key] for r in rows], dtype=float)
        values.append(var[np.isfinite(var) & (var > 0)])
        ax.semilogy(z, var, linestyle=linestyles[i % len(linestyles)],
                    marker=markers[i % len(markers)], linewidth=2, markersize=5,
                    label=res["label"])
    ax.set_xlabel("Redshift $z$")
    ax.set_ylabel(ylabel)
    ax.set_title(Path(fname).stem)
    ax.grid(True, which="both", ls="--", alpha=0.4)
    ax.legend(fontsize=9)
    ylim = robust_ylim(values)
    if ylim is not None:
        ax.set_ylim(*ylim)
    fig.tight_layout()
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    fig.savefig(OUT_DIR / fname, dpi=150)
    plt.show()
    plt.close(fig)


def plot_variance_curves(results, mock_curve_list):
    """方差图：realspace(signal/noise/analytic) + mock(k 空间, z 对齐)。"""
    for res in results:
        if "variance_table" not in res:
            raise KeyError(
                f"{res['label']} 缺少 variance_table（需先 compute_realspace_variance_table）"
            )
    _plot_variance_vs_z(results, "var_signal_mK2",
                        r"Signal variance  $\sigma^2_{\rm sig}$  [mK$^2$]",
                        "variance_signal.png")
    _plot_variance_vs_z(results, "var_noise_mK2",
                        r"Noise variance  $\sigma^2_{\rm noise}$  [mK$^2$]",
                        "variance_noise.png")
    _plot_variance_vs_z(results, "var_total_mK2",
                        r"Total variance  $\sigma^2_{\rm sig}+\sigma^2_{\rm noise}$  [mK$^2$]",
                        "variance_analytic.png")
    if mock_curve_list:
        lab = []
        for c in mock_curve_list:
            if isinstance(c, tuple):
                lab.append(c)
            else:
                cc = dict(c)
                cc["label"] = f"mock z={cc.get('z_ref', float('nan')):.2f}"
                lab.append((cc["label"], cc))
        _plot_curves(lab, "Var_mock",
                     r"Mock variance  $f_{\rm arr}P_{\rm HI}^2$  [mK$^4$ Mpc$^6$]",
                     "variance_mock.png")
        _plot_curves(lab, "Var_tot_mock",
                     r"Analytic+mock variance  $(P_{\rm HI}+P_{\rm noise})^2+f_{\rm arr}P_{\rm HI}^2$"
                     r"  [mK$^4$ Mpc$^6$]", "variance_analytic_plus_mock.png")

def plot_pnoise_signal_snr(results: List[dict]) -> None:
    curve_list = [(r["label"], r["curves"]) for r in results]
    _plot_curves(curve_list, "P_noise",
                 r"$P_{\rm noise}(k)$  [mK$^2$ Mpc$^3$]", "pnoise.png")
    _plot_curves(curve_list, "P_HI",
                 r"$P_{\rm HI}(k)$  [mK$^2$ Mpc$^3$]", "psignal.png")
    _plot_curves(curve_list, "SNR", r"$P_{\rm HI}/P_{\rm noise}$", "snr.png")


# ---- ④ 信号对参数偏导：每参数一张图，多望远镜同图 ----
def plot_derivatives(results: List[dict]) -> None:
    curve_list = [(r["label"], r["curves"]) for r in results]
    for p in results[0]["parameters"]:
        _plot_curves(curve_list, f"derivs.{p}",
                     rf"$\partial P_{{\rm HI}}/\partial {p}$", f"deriv_{p}.png",
                     symlog_y=True)


# ============================================================================
# 6b. realspace voxel 方差表：var_signal / var_noise / var_total（每 z-bin 一行）
#     保留自 6.7-interferometer.ipynb；survey_type → telescope（新结果字典字段）
#     - var_signal = ∫d³k/(2π)³ P_HI(k,μ,z)·W_vox²(k,μ)
#     - var_noise  = baofisher.noise_rms_per_voxel(z)²   （单碟口径，保留原实现）
#     - var_total  = var_signal + var_noise
#     用到哪个望远镜就为哪个生成并保存对应配置的方差表（CSV + joblib）。
# ============================================================================
import csv


def voxel_window_squared(
    engine: CosmologyEngine,
    expt_dict: dict,
    k_arr: np.ndarray,
    z: float,
    mu_arr: np.ndarray,
) -> Tuple[np.ndarray, dict]:
    k = np.atleast_1d(k_arr).astype(float)
    mu = np.atleast_1d(mu_arr).astype(float)
    z = float(z)

    r_Mpc = float(engine.comoving_distance_Mpc(z))
    H_z = float(engine.cosmo.Hubble(z))
    rnu_Mpc = (1.0 + z) ** 2 / H_z

    l = 3e8 * (1.0 + z) / (1e6 * expt_dict["nu_line"])
    theta_fwhm = l / expt_dict["Ddish"]

    sigma_kpar = (
        np.sqrt(16.0 * np.log(2.0))
        * expt_dict["nu_line"]
        / (expt_dict["dnu"] * rnu_Mpc)
    )
    sigma_kperp = np.sqrt(16.0 * np.log(2.0)) / (r_Mpc * theta_fwhm)

    mu2 = mu[None, :] ** 2
    B_tot = (k[:, None] ** 2) * ((1.0 - mu2) / sigma_kperp**2 + mu2 / sigma_kpar**2)
    W2 = np.exp(-B_tot)

    meta = {
        "r_Mpc": r_Mpc,
        "H_z": H_z,
        "rnu_Mpc": rnu_Mpc,
        "theta_fwhm_rad": theta_fwhm,
        "sigma_kperp_Mpc_inv": float(sigma_kperp),
        "sigma_kpar_Mpc_inv": float(sigma_kpar),
    }
    return W2, meta


def integrate_variance_kmu(
    k_arr: np.ndarray,
    mu_arr: np.ndarray,
    power_2d: np.ndarray,
) -> float:
    k = np.asarray(k_arr, dtype=float)
    mu = np.asarray(mu_arr, dtype=float)
    power = np.asarray(power_2d, dtype=float)
    power = np.where(np.isfinite(power), power, 0.0)

    mu_int = np.trapz(power, mu, axis=1)
    return float(np.trapz((k**2) * mu_int, k) / (4.0 * np.pi**2))


def compute_realspace_variance_table(result: dict) -> List[dict]:
    """每个 z-bin 一行：var_signal / var_noise / var_total（real-space voxel 口径）。"""
    engine = result["engine"]
    expt = copy.deepcopy(result["expt"])
    rows = []

    for bin_data in result["bin_data"]:
        k = np.asarray(bin_data["k"], dtype=float)
        mu = np.asarray(bin_data["mu"], dtype=float)
        z_mid = float(bin_data["z_mid"])

        W2, meta = voxel_window_squared(engine, expt, k, z_mid, mu)

        P_signal_voxel = np.asarray(bin_data["P_HI"], dtype=float) * W2
        var_signal = integrate_variance_kmu(k, mu, P_signal_voxel)

        var_noise = float(bf.noise_rms_per_voxel(z_mid, expt) ** 2)
        var_total = var_signal + var_noise

        rows.append(
            {
                "telescope": result["telescope"],
                "z_low": float(bin_data["z_low"]),
                "z_high": float(bin_data["z_high"]),
                "z_mid": z_mid,
                "k_min_Mpc_inv": float(k[0]),
                "k_max_Mpc_inv": float(k[-1]),
                "r_Mpc": float(meta["r_Mpc"]),
                "H_over_c_Mpc_inv": float(meta["H_z"]),
                "theta_fwhm_rad": float(meta["theta_fwhm_rad"]),
                "sigma_kperp_Mpc_inv": float(meta["sigma_kperp_Mpc_inv"]),
                "sigma_kpar_Mpc_inv": float(meta["sigma_kpar_Mpc_inv"]),
                "var_signal_mK2": float(var_signal),
                "var_noise_mK2": float(var_noise),
                "var_total_mK2": float(var_total),
                "rms_signal_mK": float(np.sqrt(max(var_signal, 0.0))),
                "rms_noise_mK": float(np.sqrt(max(var_noise, 0.0))),
                "rms_total_mK": float(np.sqrt(max(var_total, 0.0))),
            }
        )

    return rows


def save_variance_table(result: dict, out_dir: Path = OUT_DIR) -> dict:
    """保存单望远镜的 realspace 方差表（CSV + joblib）到 outputs/。"""
    from joblib import dump

    rows = result["variance_table"]
    label = result["telescope"].replace(" ", "_")
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    csv_path = out_dir / f"realspace_voxel_variance_planck_{label}.csv"
    joblib_path = out_dir / f"realspace_voxel_variance_planck_{label}.joblib"

    fieldnames = list(rows[0].keys())
    with csv_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)

    dump(
        {
            "telescope": result["telescope"],
            "note": (
                "Signal variance uses P_HI(k, mu, z) * W_vox^2 integrated over the notebook k-grid; "
                "noise variance uses baofisher.noise_rms_per_voxel(z)^2."
            ),
            "rows": rows,
        },
        joblib_path,
    )

    return {"csv": str(csv_path), "joblib": str(joblib_path)}


def plot_realspace_variance(results: List[dict], save_prefix: str = None) -> None:
    """3 子图（signal/noise/total）多望远镜同图：var vs z。y 轴用鲁棒窗口。"""
    import matplotlib.pyplot as plt

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))

    linestyles = ["-", "--", "-.", ":"]
    markers = ["o", "s", "^", "d"]

    for ax, var_type in zip(axes, ["signal", "noise", "total"]):
        allv = []
        for i, res in enumerate(results):
            label = res["label"]
            rows = res["variance_table"]
            z = [r["z_mid"] for r in rows]
            var = [r[f"var_{var_type}_mK2"] for r in rows]
            allv.append(np.asarray(var, dtype=float))
            ax.semilogy(
                z,
                var,
                linestyle=linestyles[i % len(linestyles)],
                marker=markers[i % len(markers)],
                label=label,
                linewidth=2,
                markersize=4,
            )
        ax.set_xlabel("Redshift $z$")
        ax.set_ylabel(f"{var_type.capitalize()} variance (mK²)")
        ax.grid(True, alpha=0.3)
        ax.legend()
        ylim = robust_ylim(allv)
        if ylim is not None:
            ax.set_ylim(*ylim)

    plt.tight_layout()
    if save_prefix:
        OUT_DIR.mkdir(parents=True, exist_ok=True)
        plt.savefig(OUT_DIR / f"{save_prefix}_realspace_variance.png", dpi=150)
    plt.show()


# ============================================================================
# 7. CMB Fisher + 约束表
# ============================================================================
def planck_fisher_matrix_for_params(params: List[str]) -> np.ndarray:
    F_cmb = np.array(
        [
            [1.3814e04, -3.2729e01, 1.8767e05, 4.9029e02, 6.7237e03, -9.1151e03, 0.0000e00],
            [-3.2729e01, 1.1054e08, 4.1059e07, -2.3063e08, -1.4401e06, -9.1987e06, 0.0000e00],
            [1.8767e05, 4.1059e07, 1.3256e09, -2.5375e08, 1.7445e07, -2.1510e07, 0.0000e00],
            [4.9029e02, -2.3063e08, -2.5375e08, 5.0301e08, 8.3734e05, 2.1457e07, 0.0000e00],
            [6.7237e03, -1.4401e06, 1.7445e07, 8.3734e05, 5.7721e05, -1.6066e05, 0.0000e00],
            [-9.1151e03, -9.1987e06, -2.1510e07, 2.1457e07, -1.6066e05, 1.0292e06, 0.0000e00],
            [0.0000e00, 0.0000e00, 0.0000e00, 0.0000e00, 0.0000e00, 0.0000e00, 3.8447e-02],
        ]
    )
    names = ["lnA", "Omega_m", "omega_b", "omega_cdm", "n_s", "w0", "f_NL"]
    return fisher_submatrix(F_cmb, names, params)


def sigma_from_fisher(F: np.ndarray) -> np.ndarray:
    try:
        cov = np.linalg.inv(F)
    except np.linalg.LinAlgError:
        cov = np.linalg.pinv(F)
    return np.sqrt(np.abs(np.diag(cov)))


# ---- ① 每望远镜 7 列约束表 ----
def constraints_table_data(forecast: dict):
    """单望远镜 7 列约束表（行 = 7 参数）。返回 pandas DataFrame。"""
    import pandas as pd
    ana = forecast["analytic"]
    mock = forecast.get("mock")
    params = ana["parameters"]
    n = len(params)
    sig_cmb = ana["sig_cmb"]
    pct = lambda sig: (sig_cmb - sig) / np.where(sig_cmb != 0, sig_cmb, np.nan) * 100.0
    data = {
        "σ(F_ana)": ana["sig_hi"],
        "σ(F_ana+F_CMB)": ana["sig_joint"],
        "σ(F_CMB)": sig_cmb,
        "σ(F_ana+F_CMB) vs_CMB%": pct(ana["sig_joint"]),
    }
    if mock is not None:
        data["σ(F_ana+mock)"] = mock["sig_hi"]
        data["σ(F_ana+mock+F_CMB)"] = mock["sig_joint"]
        data["σ(F_ana+mock+F_CMB) vs_CMB%"] = pct(mock["sig_joint"])
    else:
        data["σ(F_ana+mock)"] = np.full(n, np.nan)
        data["σ(F_ana+mock+F_CMB)"] = np.full(n, np.nan)
        data["σ(F_ana+mock+F_CMB) vs_CMB%"] = np.full(n, np.nan)
    df = pd.DataFrame(data, index=params)
    df.index.name = "param"
    return df


def print_constraints_summary(forecast: dict) -> None:
    """打印单个望远镜的 7 列约束汇总表。"""
    import pandas as pd
    ana = forecast["analytic"]
    df = constraints_table_data(forecast)
    cfg = ana["cfg"]
    print("\n" + "=" * 92)
    print(f"Parameter constraints — {ana['label']}   (z=[{cfg.z_bins[0]:.3f}, {cfg.z_bins[1]:.3f}], "
          f"{len(ana['bin_data'])} bins)")
    print("=" * 92)
    try:
        from IPython.display import display
        styled = df.style.format("{:.6g}", na_rep="—")
        display(styled)
    except Exception:
        with pd.option_context("display.precision", 6):
            print(df.to_string(na_rep="—"))


def save_constraints_csv(forecast: dict, out_dir: Path = OUT_DIR) -> Path:
    """保存单望远镜约束表到 outputs/constraints_{telescope}.csv。"""
    out_dir.mkdir(parents=True, exist_ok=True)
    label = forecast["analytic"]["telescope"].replace(" ", "_")
    path = out_dir / f"constraints_{label}.csv"
    constraints_table_data(forecast).to_csv(path, float_format="%.6g")
    return path


# ============================================================================
# 8. 运行入口
# ============================================================================
_ENGINE_CACHE: dict = {}


def get_engine(cfg: SurveyConfig) -> CosmologyEngine:
    """按 (z_bins, kref) 缓存 classy 引擎；同 z 范围的望远镜复用。"""
    key = (round(cfg.z_bins[0], 4), round(cfg.z_bins[1], 4), cfg.kref_T_h_Mpc)
    if key not in _ENGINE_CACHE:
        _ENGINE_CACHE[key] = CosmologyEngine(config=cfg)
    return _ENGINE_CACHE[key]


def _assemble_result(telescope, cfg, expt, engine, fc, z_edges, bin_data, F_hi, tag: str):
    """把 HI Fisher 与 CMB Fisher 联合，得到单套约束结果。tag ∈ {"analytic", "mock"}。"""
    F_cmb = planck_fisher_matrix_for_params(fc.parameters)
    F_joint = F_hi + F_cmb
    sig_hi = sigma_from_fisher(F_hi)
    sig_cmb = sigma_from_fisher(F_cmb)
    sig_joint = sigma_from_fisher(F_joint)
    return {
        "label": telescope if tag == "analytic" else f"{telescope}_mock",
        "telescope": telescope,
        "tag": tag,
        "use_mock": tag == "mock",
        "cfg": cfg, "expt": expt, "engine": engine, "forecast": fc,
        "z_edges": z_edges, "bin_data": bin_data,
        "parameters": list(fc.parameters),
        "F_hi": F_hi, "F_cmb": F_cmb, "F_joint": F_joint,
        "sig_hi": sig_hi, "sig_cmb": sig_cmb, "sig_joint": sig_joint,
    }


def run_forecast(telescope: str) -> dict:
    """解析 Fisher +（可选）按 z 插值对齐叠加 mock 方差。"""
    cfg = SurveyConfig(telescope=telescope)
    expt = cfg.load()
    engine = get_engine(cfg)
    fc = HIFisherForecast(engine, cfg)
    cache = fc.compute_bin_data(expt, cfg.include_Cfg)

    F_ana, z_edges, bin_data = fc.fisher_sum_analytic(cache)
    out = {"analytic": _assemble_result(telescope, cfg, expt, engine, fc,
                                        z_edges, bin_data, F_ana, "analytic")}
    idx = out["analytic"]["parameters"].index("f_NL")
    print(
        f"[analytic] {telescope:>24s} | z=[{cfg.z_bins[0]:.3f},{cfg.z_bins[1]:.3f}] "
        f"({len(z_edges)-1} bins) | σ(f_NL) HI={out['analytic']['sig_hi'][idx]:.6g} "
        f"Joint={out['analytic']['sig_joint'][idx]:.6g}"
    )

    mock_file = TELESCOPES[telescope].mock_file
    if mock_file is not None:
        import joblib
        mock_list = joblib.load(mock_file)

        # 红移对齐检查（打印，便于核对 mock z 中心假设）
        zm = [round(float(bd["z_mid"]), 3) for bd in bin_data]
        print(f"[mock]    {telescope:>24s} | 解析 z_mid={zm}")
        print(f"[mock]    {'':>24s} | mock z     ={[round(float(v),3) for v in MOCK_Z_CENTERS]}")

        F_mock, _, _ = fc.fisher_sum_mock_zinterp(cache, mock_list)
        out["mock"] = _assemble_result(telescope, cfg, expt, engine, fc,
                                       z_edges, bin_data, F_mock, "mock")
        out["mock_list"] = mock_list
        print(f"[mock]    {telescope:>24s} | 解析+mock(z对齐) | σ(f_NL) HI={out['mock']['sig_hi'][idx]:.6g} "
              f"Joint={out['mock']['sig_joint'][idx]:.6g}")
    return out



# ============================================================================
# 9. main：望远镜列表驱动 + 规范输出
# ============================================================================
# 想跑哪个望远镜就写进这里（键见 TELESCOPES，命名 {telescope}_{mode}）。
# 输出：
#   ① 每望远镜一张 7 列约束表 + outputs/constraints_{telescope}.csv
#   ② 5 张方差图（variance_signal/noise/analytic=realspace 方差 σ² vs z；mock=k 空间 mock 方差）
#   ③ 3 张 P_noise/P_signal/SNR 图（pnoise/psignal/snr.png）
#   ④ 7 张偏导图（deriv_{param}.png）
#   ⑤ 每望远镜一张 realspace voxel 方差表（var_signal/var_noise/var_total，每 z-bin 一行）
#      + outputs/realspace_voxel_variance_planck_{telescope}.csv/.joblib
#      + voxel_realspace_variance.png（3 子图：signal/noise/total vs z）
# 给某个条目配 mock：在 TELESCOPES 里加 mock_file="xxx.joblib"（不绑定干涉/单碟）。
SURVEYS = ["ska_mid1_interferom", "chime_icyl", "hirax_interferom", "tianlai_icyl","cv_z0to3_interferom","meerkat_b1_interferom"]


def main():
    forecasts = [run_forecast(t) for t in SURVEYS]

    # ① 约束表（每望远镜一张 7 列）+ CSV
    for f in forecasts:
        print_constraints_summary(f)
        path = save_constraints_csv(f)
        print(f"[saved]   {path}")

    # ② 组装 k 空间曲线 + realspace 方差表（方差图要用方差数据，先算好）
    ana = [f["analytic"] for f in forecasts]
    for res in ana:
        build_curves(res)
        res["variance_table"] = compute_realspace_variance_table(res)
    mock_curves = [c for c in (build_mock_curves(f) for f in forecasts) if c is not None]

    # ③-⑤ 图
    plot_variance_curves(ana, mock_curves)
    plot_pnoise_signal_snr(ana)
    plot_derivatives(ana)

    # ⑥ 保存方差表（CSV + joblib）+ 3 子图汇总
    for res in ana:
        paths = save_variance_table(res)
        print(f"[saved]   {paths['csv']}")
    plot_realspace_variance(ana, save_prefix="voxel")


# 唯一执行入口：解析 + mock(z 对齐) 全套结果
main()
```

## 诊断：Var_mock 与解析方差的大小比较

上方主 cell 的 mock 版 Fisher 中，每个 (k, μ) 模式的总方差为

$$\mathrm{Var}_{\rm tot} = \underbrace{\big(P_{\rm HI}+P_{\rm noise}\big)^2}_{\mathrm{Var}_{\rm ana}\,(=\mathrm{tot\_safe}^2)} \;+\; \underbrace{f_{\rm arr}\!\cdot\!P_{\rm HI}^{\,2}}_{\mathrm{Var}_{\rm mock}}$$

本诊断 cell 在参考红移 bin 处对比这两项（及比值），以回答 **mock 额外方差相对解析高斯方差占多大比例、在哪些 (k, μ) 主导**：

- 前景楔 / 无基线区（$P_{\rm noise}=\infty$）的模式在 Fisher 中已被剔除，图中置空（白）；
- mock 网格外 $f_{\rm arr}=0 \Rightarrow \mathrm{Var}_{\rm mock}=0$，即该模式退回纯解析；
- 判据：$\mathrm{ratio}=\mathrm{Var}_{\rm mock}/\mathrm{Var}_{\rm ana}$，ratio < 1 表示 mock 只是次要修正，≥ 1 表示 mock 主导。


```python
# ============================================================================
#  Diagnostic cell: Var_mock = f_arr * P_HI^2  vs  Var_ana = (P_HI + P_noise)^2
#  Reuses the main cell namespace (SurveyConfig/TELESCOPES/MOCK_FILE/
#  MOCK_Z_CENTERS/get_engine/HIFisherForecast/BF_INF_NOISE/robust_ylim/OUT_DIR)
#  and follows the exact same convention as fisher_sum_mock_zinterp.
#  If NameError: run the main cell first.
#  NOTE: keep figure text in English/math only (no CJK glyphs in DejaVu fonts).
# ============================================================================
import numpy as np
import matplotlib as _mpl
_mpl.rcParams["text.usetex"] = False
import matplotlib.pyplot as plt
import joblib
from scipy.interpolate import RegularGridInterpolator

_required = ["SurveyConfig", "TELESCOPES", "MOCK_FILE", "MOCK_Z_CENTERS",
             "get_engine", "HIFisherForecast", "BF_INF_NOISE",
             "robust_ylim", "OUT_DIR"]
_missing = [n for n in _required if n not in globals()]
if _missing:
    raise RuntimeError("Missing main-cell namespace: " + ", ".join(_missing)
                       + "\nRun the main cell (all definitions + main()) first.")

DIAG_TELS = ["ska_mid1_interferom"]   # <- telescopes to inspect (keys in TELESCOPES)
MUS_RATIO = [0.0, 0.3, 0.6, 0.9]      # mu values for the ratio curves
MU_REF    = 0.5                       # fixed mu for the main comparison panel

# ---- load mock (6 layers share one (k_perp,k_par) grid; same rule as fisher_sum_mock_zinterp)
mock_list = joblib.load(MOCK_FILE)
z_mock = np.asarray(MOCK_Z_CENTERS, dtype=float)
_kp0, _kp1 = mock_list[0]["kper_bin"], mock_list[0]["kpar_bin"]
_cube = np.stack([m["fisher_arr"] for m in mock_list], axis=0)   # (nz,200,200)
_mock3d = RegularGridInterpolator((z_mock, _kp0, _kp1), _cube,
                                  bounds_error=False, fill_value=0.0)


def _ref_z_bin_index(cfg):
    """Bin z exactly like compute_bin_data; return (edges, central-bin index)."""
    zmin, zmax = cfg.z_bins
    dz = float(cfg.delta_z)
    ze = np.linspace(zmin, zmax, int(round((zmax - zmin) / dz)) + 1)
    ze[-1] = zmax
    return ze, (len(ze) - 2) // 2


for TEL in DIAG_TELS:
    cfg = SurveyConfig(telescope=TEL)
    expt = cfg.load()
    engine = get_engine(cfg)

    z_edges, bi = _ref_z_bin_index(cfg)
    z_low, z_high = float(z_edges[bi]), float(z_edges[bi + 1])
    z_mid = 0.5 * (z_low + z_high)

    ks = np.logspace(np.log10(cfg.k_min), np.log10(cfg.k_max), int(cfg.nk))
    mu = np.linspace(-1.0, 1.0, int(cfg.n_mu))

    P_HI    = engine.HI_power_spectrum(ks, z_mid, mu)                   # (nk,nmu)
    P_noise = engine.P_noise_grid(ks, mu, z_mid, z_low, z_high,
                                  expt, cfg.include_Cfg)                # (nk,nmu)

    # --- interpolate mock f_arr to this bin ((k,mu) -> (k_perp,k_par), out-of-range = 0)
    kper = ks[:, None] * np.sqrt(np.clip(1.0 - mu[None, :] ** 2, 0.0, None))
    kpar = ks[:, None] * np.abs(mu[None, :])
    zc = float(np.clip(z_mid, z_mock[0], z_mock[-1]))
    _npts = ks.size * mu.size
    _pts = np.stack([np.full(_npts, zc),
                     np.broadcast_to(kper, P_HI.shape).ravel(),
                     np.broadcast_to(kpar, P_HI.shape).ravel()], axis=-1)
    f_arr = _mock3d(_pts).reshape(P_HI.shape)

    # --- the two variance terms (same units mK^4 Mpc^6; tot_safe = P_HI + P_noise)
    Var_ana  = (P_HI + P_noise) ** 2
    Var_mock = f_arr * P_HI ** 2
    masked   = ~np.isfinite(P_noise)          # wedge / no-baseline: dropped in Fisher

    # --- statistics
    ok    = np.isfinite(Var_ana) & (Var_ana > 0) & (Var_mock > 0) & (~masked)
    ratio = np.where(ok, Var_mock / Var_ana, np.nan)
    n_over = int(np.sum(ok & (ratio >= 1.0)))
    n_near = int(np.sum(ok & (ratio >= 0.1)))
    n_f0   = int(np.sum((~masked) & (f_arr == 0.0)))
    n_mask = int(np.sum(masked))
    print("=" * 76)
    print(f"[{TEL}] z-bin {bi}: z_mid={z_mid:.3f}  "
          f"(mock z coverage [{z_mock[0]:.2f},{z_mock[-1]:.2f}])")
    print(f"  valid modes {int(ok.sum())}/{ks.size*mu.size} | "
          f"wedge/no-baseline dropped {n_mask} | mock not covered (f=0) {n_f0}")
    rv = ratio[ok]
    if rv.size:
        print("  ratio = Var_mock/Var_ana : "
              f"median={np.median(rv):.3e}  p90={np.percentile(rv, 90):.3e}  "
              f"p99={np.percentile(rv, 99):.3e}  max={np.max(rv):.3e}")
        print(f"  mock-dominated (ratio>=1): {n_over} modes | "
              f"non-negligible (ratio>=0.1): {n_near} modes")

    # ---------- Panel 1: Var_mock vs Var_ana (fixed mu=MU_REF; thin lines = other mu)
    fig, ax = plt.subplots(figsize=(9, 6))
    iu = int(np.argmin(np.abs(mu - MU_REF)))
    for j in range(mu.size):
        ax.loglog(ks, np.where(~masked[:, j], Var_ana[:, j], np.nan),
                  color="tab:blue", alpha=0.05, lw=1)
        ax.loglog(ks, np.where((~masked[:, j]) & (Var_mock[:, j] > 0),
                               Var_mock[:, j], np.nan),
                  color="tab:red", alpha=0.05, lw=1)
    ax.loglog(ks, np.where(~masked[:, iu], Var_ana[:, iu], np.nan),
              "-", color="tab:blue", lw=2.5,
              label=r"${\rm Var}_{\rm ana}=(P_{\rm HI}+P_{\rm noise})^2$")
    ax.loglog(ks, np.where((~masked[:, iu]) & (Var_mock[:, iu] > 0),
                           Var_mock[:, iu], np.nan),
              "--", color="tab:red", lw=2.5,
              label=r"${\rm Var}_{\rm mock}=f_{\rm arr}P_{\rm HI}^2$")
    ax.loglog(ks, P_HI[:, iu] ** 2, ":", color="gray", lw=1.5,
              label=r"$P_{\rm HI}^2$ (CV limit)")
    ax.set_xlabel(r"$k$ [Mpc$^{-1}$]")
    ax.set_ylabel(r"variance  [mK$^4$ Mpc$^6$]")
    ax.set_title(f"{TEL}: Var_mock vs Var_ana  (z={z_mid:.2f}, "
                 r"$\mu$" + f"={mu[iu]:.2f}; thin lines: other $\mu$)")
    ax.grid(True, which="both", ls="--", alpha=0.35)
    ax.legend(fontsize=10)
    fig.tight_layout()
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    fig.savefig(OUT_DIR / f"diag_mock_vs_ana_{TEL}.png", dpi=150)
    plt.show()
    plt.close(fig)

    # ---------- Panel 2: ratio = Var_mock/Var_ana (several mu, y=1 reference)
    fig, ax = plt.subplots(figsize=(9, 6))
    _ls, _vals = ["-", "--", "-.", ":"], []
    for jt, mut in enumerate(MUS_RATIO):
        iq = int(np.argmin(np.abs(mu - mut)))
        y = np.where(ok[:, iq], ratio[:, iq], np.nan)
        ax.semilogy(ks, y, _ls[jt % 4], lw=2, label=rf"$\mu$={mu[iq]:.2f}")
        _vals.append(y[np.isfinite(y) & (y > 0)])
    ax.axhline(1.0, color="k", ls=":", lw=1.2)
    ax.text(ks[1], 1.6, "mock term = analytic term (ratio=1)", fontsize=9)
    ax.set_xscale("log")
    ax.set_xlabel(r"$k$ [Mpc$^{-1}$]")
    ax.set_ylabel(r"${\rm Var}_{\rm mock}/{\rm Var}_{\rm ana}$")
    ax.set_title(f"{TEL}: Var_mock/Var_ana ratio  (z={z_mid:.2f})")
    ax.grid(True, which="both", ls="--", alpha=0.35)
    _yl = robust_ylim(_vals)
    if _yl is not None:
        ax.set_ylim(min(_yl[0], 1e-4), max(_yl[1], 10.0))
    ax.legend(fontsize=9)
    fig.tight_layout()
    fig.savefig(OUT_DIR / f"diag_mock_over_ana_{TEL}.png", dpi=150)
    plt.show()
    plt.close(fig)

    # ---------- Panel 3: ratio map over (k, mu>=0)
    mupos = mu[mu >= 0.0]
    ipos  = np.where(mu >= 0.0)[0]
    Rlog  = np.full((ks.size, mupos.size), np.nan)
    for jj, j in enumerate(ipos):
        Rlog[:, jj] = np.where(ok[:, j], np.log10(ratio[:, j]), np.nan)
    fig, ax = plt.subplots(figsize=(9, 6))
    im = ax.pcolormesh(ks, mupos, Rlog.T, cmap="RdYlBu_r",
                       vmin=-4.0, vmax=0.5, shading="auto")
    cb = fig.colorbar(im, ax=ax)
    cb.set_label(r"$\log_{10}({\rm Var}_{\rm mock}/{\rm Var}_{\rm ana})$")
    ax.set_xscale("log")
    ax.set_xlabel(r"$k$ [Mpc$^{-1}$]")
    ax.set_ylabel(r"$\mu$")
    ax.set_title(f"{TEL}: Var_mock/Var_ana ratio map (z={z_mid:.2f})  "
                 "(blue: mock small, red: mock large, white: excluded)")
    fig.tight_layout()
    fig.savefig(OUT_DIR / f"diag_mock_ratio2d_{TEL}.png", dpi=150)
    plt.show()
    plt.close(fig)

print("done")
```

```python
# ============================================================================
#  Strict calculation: SKA (ska_mid1 = SKA1MIDbase1) single-dish vs
#  interferometer noise. SAME params (ttot = 1e5 h, dnu, Sarea, Tsys(z),
#  Ddish, Ndish); only expt['mode'] differs (+ n(u) file for interferom).
#  Self-contained; replicates notebook pipeline (KMIN/KMAX, SurveyConfig
#  rules, classy engine, P_noise = bf.Cnoise(...)*r^2*rnu).
#  (A) convenience formulas -> reference only (different voxel units).
#  (B) strict Fisher metric P_noise(k,mu) used by fisher_sum_*.
# ============================================================================
import sys, io, copy, contextlib
from pathlib import Path

import matplotlib as _mpl
_mpl.rcParams["text.usetex"] = False     # 关闭 LaTeX（缺 type1cm.sty）
import matplotlib.pyplot as plt
import numpy as np

BAO21CM_ROOT = "/home/liuhk/bao21cm-master/bao21cm-master"
if BAO21CM_ROOT not in sys.path:
    sys.path.insert(0, BAO21CM_ROOT)
from radiofisher import baofisher_k as bf
from radiofisher import experiments as ex
from classy import Class

bf.KMIN_EFF = 5.0e-3
bf.KMAX_EFF = 0.535
BF_INF = float(bf.INF_NOISE)
T_OBS_H = 1.0e4        # 与 SurveyConfig 默认一致（10^5 h）


def assemble(mode):
    spec = copy.deepcopy(ex.SKA1MIDbase1)
    spec["mode"] = mode
    spec["ttot"] = T_OBS_H * 3600 * 1e6
    if mode[0] == "i" and isinstance(spec.get("n(x)"), str):
        p = Path(BAO21CM_ROOT) / spec["n(x)"]
        if p.exists():
            spec["n(x)"] = bf.load_interferom_file(str(p))
    if not callable(spec.get("n(x)")):
        spec.pop("n(x)", None)
    return spec


def engine(zmax_pk):
    c = Class()
    c.set({"output": "mPk", "P_k_max_h/Mpc": 5.0, "non linear": "halofit",
           "z_max_pk": float(zmax_pk),
           "h": float(np.sqrt((0.02237 + 0.1200) / 0.3153)),
           "omega_b": 0.02237, "omega_cdm": 0.1200,
           "Omega_Lambda": 1.0 - 0.3153, "A_s": 2.1e-9, "n_s": 0.9649,
           "w0_fld": -1.0, "m_ncdm": "0.06", "N_ncdm": "1"})
    c.compute()
    return c


def p_noise(c, ks, mu, z, zl, zh, expt):
    r = float(c.comoving_distance(z)); H = float(c.Hubble(z))
    rnu = (1.0 + z) ** 2 / H
    k2 = np.atleast_2d(ks).T; m2 = np.atleast_2d(mu)
    q = r * k2 * np.sqrt(np.clip(1.0 - m2 ** 2, 0, None))
    y = rnu * k2 * np.abs(m2)
    cd = {"z": float(z), "aperp": 1.0, "apar": 1.0, "r": r, "rnu": rnu, "ns": 0.9649}
    e = copy.deepcopy(expt)
    e["dnutot"] = max(float(expt["nu_line"] / (1.0 + zl)
                            - expt["nu_line"] / (1.0 + zh)), 1e-12)
    with contextlib.redirect_stdout(io.StringIO()):
        cn = bf.Cnoise(q, y, cd, e, cv=False)
    bad = ~np.isfinite(cn) | (cn >= 0.5 * BF_INF)
    return np.where(bad, np.inf, cn * r ** 2 * rnu)


exptD = assemble("dish")
exptI = assemble("interferom")
numax, dnutot = exptI["survey_numax"], exptI["survey_dnutot"]
nu_line = exptI["nu_line"]
zmin = nu_line / numax - 1.1028
zmax = nu_line / (numax - dnutot) - 0.8083
zc = np.arange(zmin + 0.25, zmax - 1e-9, 0.5)
edges = np.concatenate([[zmin], zc + 0.25, [zmax]]) if zc[-1] + 0.25 < zmax - 1e-9 \
    else np.linspace(zmin, zmax, len(zc) + 1)

cosmo = engine(zmax + 0.5)
mu = np.linspace(-1, 1, 100)
ks = np.logspace(np.log10(5e-3), np.log10(0.535), 50)
imu = np.argmin(np.abs(mu - 0.5)); ik05 = np.argmin(np.abs(ks - 0.05))

print("SKA1MIDbase1: Ndish=190 Ddish=15 m dnu=0.1 MHz "
      f"Sarea={exptI['Sarea']:.4f} sr  ttot={T_OBS_H:g} h")
print(f"{'z':>4} | {'sig_d^2(A)':>9} {'sig_i^2(A)':>10} {'aux_i/d':>8} | "
      f"{'P_d@.05':>9} {'P_i@.05':>10} {'i/d@.05':>8} | {'med_i/d':>8} {'p90':>6}")
allPd, allPi, allz = [], [], []
for i in range(len(zc)):
    z, zl, zh = float(zc[i]), float(edges[i]), float(edges[i + 1])
    sd2 = bf.noise_rms_per_voxel(z, exptD) ** 2
    si2 = bf.noise_rms_per_voxel_interferom(z, exptI) ** 2
    Pd = p_noise(cosmo, ks, mu, z, zl, zh, exptD)
    Pi = p_noise(cosmo, ks, mu, z, zl, zh, exptI)
    vd, vi = np.isfinite(Pd) & (Pd > 0), np.isfinite(Pi) & (Pi > 0)
    both = vd & vi
    with np.errstate(divide="ignore", invalid="ignore"):
        r = np.where(both, Pi / Pd, np.nan)[both]
    med = float(np.median(r)) if r.size else np.nan
    p90 = float(np.percentile(r, 90)) if r.size else np.nan
    pdi, pii = Pd[ik05, imu], Pi[ik05, imu]
    r05 = (pii / pdi) if (np.isfinite(pdi) and np.isfinite(pii)) else np.nan
    allPd.append(Pd[:, imu]); allPi.append(Pi[:, imu]); allz.append(z)
    print(f"{z:4.1f} | {sd2:9.1e} {si2:10.1e} {si2/sd2:8.1e} | "
          f"{pdi:9.1e} {pii:10.1e} {r05:8.1e} | {med:8.1e} {p90:6.1e}  "
          f"[int {int(vi.sum())}/{int(vd.sum())}]")
print("(A) aux formulas have different voxel units -> NOT comparable.")
print("(B) P_noise rows = strict Fisher metric actually used.")

fig, ax = plt.subplots(figsize=(9, 6))
for j, i in enumerate([k for k in (1, 2, 4) if k < len(allz)]):
    ax.loglog(ks, np.where(np.isfinite(allPd[i]), allPd[i], np.nan),
              ["-", "--", "-.", ":"][j], color="tab:blue", lw=2,
              label=f"dish z={allz[i]:.2f}")
    ax.loglog(ks, np.where(np.isfinite(allPi[i]), allPi[i], np.nan),
              ["-", "--", "-.", ":"][j], color="tab:red", lw=2,
              label=f"interferom z={allz[i]:.2f}")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel(r"$k$ [Mpc$^{-1}$]")
ax.set_ylabel(r"$P_{\rm noise}(k,\mu\simeq0.5)$  [mK$^2$ Mpc$^3$]")
ax.set_title("ska_mid1: dish vs interferom P_noise (same params, mu~0.5)")
ax.grid(True, which="both", ls="--", alpha=0.4)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
print("done")
```
